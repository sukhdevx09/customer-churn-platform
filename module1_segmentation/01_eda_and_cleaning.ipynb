```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")

print("Libraries loaded successfully")
```

```python
df = pd.read_csv('../data/telco_churn.csv')
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
}).query('`Missing Count` > 0')

print("Columns with missing values:")
print(missing_df)
```

```python
churn_counts = df['Churn'].value_counts()
churn_pct = df['Churn'].value_counts(normalize=True) * 100

print("Churn Distribution:")
print(pd.DataFrame({'Count': churn_counts, 'Percentage': churn_pct.round(2)}))

# Plot
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Churn', palette=['#2ecc71','#e74c3c'])
plt.title('Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print("Categorical Columns:")
for col in categorical_cols:
    print(f"\n{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts().head())
```

```python
numerical_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()
print("Numerical Columns:", numerical_cols)

df[numerical_cols].hist(figsize=(12,6), bins=30, color='steelblue', edgecolor='white')
plt.suptitle('Distribution of Numerical Columns', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(data=df, x='tenure', hue='Churn', bins=30, palette=['#2ecc71','#e74c3c'])
plt.title('Tenure Distribution by Churn')
plt.xlabel('Tenure (Months)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=df, x='Churn', y='MonthlyCharges', palette=['#2ecc71','#e74c3c'])
plt.title('Monthly Charges by Churn')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
corr = df[numerical_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Check the data type
print("TotalCharges dtype:", df['TotalCharges'].dtype)

# Find blank/space values
blank_mask = df['TotalCharges'].str.strip() == ''
print(f"\nBlank TotalCharges rows: {blank_mask.sum()}")

# Show those rows
print(df[blank_mask][['customerID','tenure','MonthlyCharges','TotalCharges']])
```

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Duplicate customerIDs: {df['customerID'].duplicated().sum()}")
print(f"\nTotal unique customers: {df['customerID'].nunique()}")
```

```python
print("SeniorCitizen value counts:")
print(df['SeniorCitizen'].value_counts())
print("\nNote: 0 = Not Senior, 1 = Senior — stored as int but is categorical")

plt.figure(figsize=(5,3))
sns.countplot(data=df, x='SeniorCitizen', hue='Churn', palette=['#2ecc71','#e74c3c'])
plt.title('Churn by Senior Citizen Status')
plt.xticks([0,1], ['Not Senior','Senior'])
plt.tight_layout()
plt.show()
```

```python
contract_churn = df.groupby('Contract')['Churn'].value_counts(normalize=True).unstack() * 100

plt.figure(figsize=(7,4))
contract_churn['Yes'].sort_values(ascending=False).plot(kind='bar', color='#e74c3c', edgecolor='white')
plt.title('Churn Rate by Contract Type (%)')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate %')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(contract_churn.round(2))
```

```python
payment_churn = df.groupby('PaymentMethod')['Churn'].value_counts(normalize=True).unstack() * 100

plt.figure(figsize=(8,4))
payment_churn['Yes'].sort_values(ascending=False).plot(kind='bar', color='#e74c3c', edgecolor='white')
plt.title('Churn Rate by Payment Method (%)')
plt.xlabel('Payment Method')
plt.ylabel('Churn Rate %')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
internet_churn = df.groupby('InternetService')['Churn'].value_counts(normalize=True).unstack() * 100

plt.figure(figsize=(6,4))
internet_churn['Yes'].sort_values(ascending=False).plot(kind='bar', color='#e74c3c', edgecolor='white')
plt.title('Churn Rate by Internet Service (%)')
plt.xlabel('Internet Service')
plt.ylabel('Churn Rate %')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['tenure_group'] = pd.cut(df['tenure'], 
                             bins=[0,12,24,48,72], 
                             labels=['0-12 months','13-24 months','25-48 months','49-72 months'])

tenure_churn = df.groupby('tenure_group', observed=True)['Churn'].value_counts(normalize=True).unstack() * 100

plt.figure(figsize=(7,4))
tenure_churn['Yes'].plot(kind='bar', color='#e74c3c', edgecolor='white')
plt.title('Churn Rate by Tenure Group (%)')
plt.xlabel('Tenure Group')
plt.ylabel('Churn Rate %')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(tenure_churn.round(2))
```

## Key Observations from EDA

1. Dataset: 7,043 customers, 21 columns, no duplicate customerIDs
2. Churn rate: ~26% — class imbalance exists, will handle in modeling
3. TotalCharges: stored as string, 11 blank rows found — will fix in cleaning phase
4. SeniorCitizen: stored as integer but is categorical
5. Contract type: Month-to-month customers churn at ~42% vs 2-3% for long-term contracts
6. Payment method: Electronic check users churn the most
7. Internet service: Fiber optic customers churn significantly more than DSL
8. Tenure: Customers in first 12 months churn at highest rate — early retention is critical
9. Higher monthly charges + short tenure = highest churn risk
